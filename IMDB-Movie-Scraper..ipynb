```python
def get_details(soop):
  title=[]
  titles=soop.find_all('h3',class_='ipc-title__text')
  title_regex=re.compile(r'(?<=\d\.\s).*')
  for i in titles:

    extracted_title=title_regex.findall(i.text)

    title.extend(extracted_title)


  durations=soop.find_all('span',class_="sc-b189961a-8 hCbzGp cli-title-metadata-item")
  print(durations[0])
  year = []
  duration = []
  grade = []
  ratings=[]
  view = []
  year_regex = re.compile(r'\d{4}')
  duration_regex = re.compile(r'\d+h\s\d+m')
  grade_regex = re.compile(r'^[A-Z]+.*')
  extracted_year = year_regex.findall(durations[0].text)
  extracted_duration = duration_regex.findall(durations[1].text)
  extracted_grade = grade_regex.findall(durations[2].text)

  if extracted_year:
      year.extend(extracted_year)
  if extracted_duration:
      duration.extend(extracted_duration)
  if extracted_grade:
      grade.extend(extracted_grade)

  rating=soop.find_all('span',class_="ipc-rating-star--rating")

  for i in rating:
    ratings.append(i.text)

  views=soop.find_all('span',class_="ipc-rating-star--voteCount")
  for i in views:
    view.append(i.text.replace('(','').replace(')',''))

  return title[0],year[0],duration[0],grade[0],ratings[0],view[0]
```

```python
url='https://www.imdb.com/chart/top/'
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_11_5) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/50.0.2661.102 Safari/537.36'}
def get_soop(url,headers):
  response=requests.get(url,headers=headers)
  if response.status_code == 200:
    return bs(response.text)
  else:
    return None


def get_movie(url,headers):
  'https://www.imdb.com/chart/top/'
  soop=get_soop(url,headers)
  if soop:
    movie_tag=soop.find_all('li',class_='ipc-metadata-list-summary-item sc-10233bc-0 TwzGn cli-parent')
    movie=[]
    print(movie_tag)
    for movie_t in movie_tag:
      print(get_details(movie_t))
      movie.append(get_details(movie_t))
    return movie

  else:
    return None


```
